## Imports

```python
import os
from astropy.table import Table
from astropy.coordinates import SkyCoord

from astroquery.mast import Catalogs, MastMissions, Observations
from mast_table import MastTable
```

## Observation query results from `astroquery.mast.MastMissions`

### JWST observations

By running the following cell, we retrieve MAST data on [Messier 4 (M4)](https://science.nasa.gov/mission/hubble/science/explore-the-night-sky/hubble-messier-catalog/messier-4/) from the [James Webb Space Telescope (JWST)](https://www.stsci.edu/jwst). We limit our query to the first 100 results, and then we load the data into `mast-table`, which we will display inline below the cell.

```python
# Create MastMissions object to search for JWST datasets
mastmissions = MastMissions(mission='jwst')

# Query for the first 100 results on M4
jwst_observations = mastmissions.query_object(
    "M4",
    limit=100
)

# Create a mast-table instance containing the results, display inline
jwst_table = MastTable(jwst_observations)
jwst_table
```

### Roman observations

By running the following cell, we retrieve MAST data from the [Roman Space Telescope](https://www.stsci.edu/roman). We then load the data into `mast-table`and display it inline below the cell.

<div class="alert alert-warning">
Warning: the following cell relies on Roman data. If a user lacks the proper data permissions, this cell will NOT run. </br> 
Be sure to update `coordinates` with a target of your choice!
</div>

```python
# Configure MastMissions to query Roman data
mast = MastMissions(mission='roman')

# Update to use your own token from:
# https://auth.mast.stsci.edu/token?suggested_name=Astroquery&suggested_scope=mast:exclusive_access
user_token = "insert_your_token_here"
my_session = mast.login(token=user_token)

# Query for Roman WFI data on a target of your choice
valid_ra, valid_dec = (None,None)
roman_observations = mast.query_region(
    coordinates=SkyCoord(ra=valid_ra, dec=valid_dec, unit='deg'),
    radius='10 arcmin',
    exposure_type='WFI_IMAGE'
)

# Create a mast-table instance containing the results, display inline
roman_table = MastTable(roman_observations)
roman_table.show()
```

## Query results from `astroquery.mast.Catalogs`

By running the following cell, we retrieve MAST data on TRAPPIST-1 from Panoramic Survey Telescope and Rapid Response System (Pan-STARRS) Data Release 2 (DR2) and load the data into `mast-table`. We then display the `mast-table` inline below the cell.

<div class="alert alert-warning">
Warning: this query relies on the <a href="https://github.com/snbianco/astroquery/tree/mast-beta-release">astroquery:mast-beta-release</a> branch available on the Roman Research Nexus (RRN). To install this version, you can run: <br>
    &emsp;!pip install git+https://github.com/snbianco/astroquery.git@mast-beta-release<br>
Then, restart your kernel to ensure the right version is available when you run the following cell.
</div>

```python
# Query near TRAPPIST-1 for Pan-STARTS DR2 
target = SkyCoord.from_name('TRAPPIST-1')
catalog_data = Catalogs.query_region(
    target,
    radius=0.01,
    catalog="ps1_dr2",
)

# Create a mast-table instance containing the results, display inline
catalog_table = MastTable(catalog_data)
catalog_table
```

## Catalog query results from `astroquery.mast.Observations`

By running the following cell, we retrieve MAST data on [Kepler-444](https://www.nasa.gov/image-article/kepler-444-an-ancient-system-with-five-planets/) from [Galaxy Evolution Explorer (GALEX)](https://archive.stsci.edu/missions-and-data/galex) and then load the data into `mast-table`. We display the `mast-table` inline below the cell.

```python
# Query for GALEX Kepler-444 data
obs_data = Observations.query_criteria(
    objectname="Kepler-444",
    radius=0.01,
    obs_collection='GALEX'
)

# Create a mast-table instance containing the results, display inline
obs_table = MastTable(obs_data)
obs_table.show()
```
